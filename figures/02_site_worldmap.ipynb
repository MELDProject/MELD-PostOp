```python
import plotly.express as px
import pandas as pd
import numpy as np
import requests
from io import BytesIO

file = '/path/to/patient_demographics.csv'
df = pd.read_csv(file)
```

```python
# find subset site / country

dfsub = df.dropna(subset=['hospital', 'Country'])[['hospital','Country']]

dfsub = dfsub.drop_duplicates()

dfsub

# Find number of site and country 
sites = np.unique(dfsub['hospital'].values)
countries = np.unique(dfsub['Country'].values)

print(f'Number of sites: {len(sites)}')
print(f'Number of countries: {len(countries)}')

df_group = dfsub.groupby(['Country'], as_index=False).count()
df_group

# Import country standard
import numpy as np

def get_country_code(row):
    code = df_template[df_template['country'] == row['Country']]['iso_alpha'].values
    # manually add the codes not on this list
    if len(code)!=1:
        if row['Country']=='UK':
            code='GBR'
        elif row['Country']=='Brazil':
            code='BRA'
        elif row['Country']=='Spain':
            code='ESP'
        elif row['Country']=='Chile':
            code='CHL'
        elif row['Country']=='Germany':
            code='DEU'
        elif row['Country']=='Romania':
            code='ROU'
        elif row['Country']=='US':
            code='USA'
        elif row['Country']=='China':
            code='CHN'
        elif row['Country']=='France':
            code='FRA'
        elif row['Country']=='Switzerland':
            code='CHE'
        elif row['Country']=='Greece':
            code='GRC'
        elif row['Country']=='Japan':
            code='JPN'
        elif row['Country']=='Finland':
            code='FIN'
        elif row['Country']=='Italy':
            code='ITA'
        elif row['Country']=='Canada':
            code='CAN'
        elif row['Country']=='India':
            code='IND'
        else:
            code = np.nan
    else:
        code = code[0]
    return code

df_template = px.data.gapminder().filter(items=['country','iso_alpha']).drop_duplicates()

df_group['country_code'] = df_group.apply(lambda row: get_country_code(row), axis=1)
df_group

# Create scatter map

fig = px.choropleth(
    df_group,
    locations="country_code",   # ISO country codes
    color="hospital",           # Column to color by
    projection="equirectangular",
    color_continuous_scale="Viridis"
)

fig.update_geos(
    showland=True,
    landcolor="lightgrey"
)

fig.update_layout(
    
    margin={"r":0,"t":0,"l":0,"b":0},
    coloraxis_colorbar=dict(
        x=0.95,  
        xanchor="left",
        len=0.80, 
        thickness=20,
        title=dict(
            text="Hospitals",  
            side="top"      
        )
    )
)

fig.show()
```
